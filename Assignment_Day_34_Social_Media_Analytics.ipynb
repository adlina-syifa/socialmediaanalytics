### Check dan Handling Kolom Reach

```python
df_sosmed[df_sosmed["Reach"] <= 0][
    ["Content_ID", "Reach", "Impressions", "Likes", "Comments", "Shares", "Clicks"]]
```

```python
check_reach = df_sosmed[df_sosmed["Reach"] < 0].copy()

check_reach["Reach_Positive"] = check_reach["Reach"].abs()

check_reach["Total_Interactions"] = (
    check_reach["Likes"]
    + check_reach["Comments"]
    + check_reach["Shares"]
    + check_reach["Saves"].fillna(0))

check_reach["ER_Recalculated"] = (
    check_reach["Total_Interactions"]
    / check_reach["Reach_Positive"])

check_reach[[  "Content_ID",
               "Reach",
               "Reach_Positive",
               "Impressions",
               "Total_Interactions",
               "ER_Recalculated",
               "Engagement_Rate_Reach"]]
```

```python
df_sosmed["Reach"] = df_sosmed["Reach"].abs()
```

### Check dan Handling Kolom Impressions

```python
df_sosmed[df_sosmed["Impressions"] <= 0]
```

```python
df_sosmed[df_sosmed["Reach"].abs() > df_sosmed["Impressions"]][["Content_ID", "Reach", "Impressions"]]
```

```python
Q1 = df_sosmed["Impressions"].quantile(0.25)
Q3 = df_sosmed["Impressions"].quantile(0.75)

IQR = Q3 - Q1
upper_limit = Q3 + 1.5 * IQR

print("Batas atas:", upper_limit)

impressions_outlier = df_sosmed[
    df_sosmed["Impressions"] > upper_limit]

impressions_outlier[
    ["Content_ID", "Reach", "Impressions", "Clicks", "CTR"]]
```

```python
check_impressions = impressions_outlier.copy()

check_impressions["Total_Interactions"] = (
    check_impressions["Likes"]
    + check_impressions["Comments"]
    + check_impressions["Shares"]
    + check_impressions["Saves"].fillna(0))

check_impressions["CTR_Recalculated"] = (
    check_impressions["Clicks"]
    / check_impressions["Impressions"])

check_impressions["ER_Impressions_Recalculated"] = (
    check_impressions["Total_Interactions"]
    / check_impressions["Impressions"])

check_impressions[[
        "Content_ID",
        "Impressions",
        "CTR",
        "CTR_Recalculated",
        "Engagement_Rate_Impressions",
        "ER_Impressions_Recalculated"]]
```

```python
check_impressions["Impressions_Adjusted"] = (
    check_impressions["Impressions"] / 5)

check_impressions["CTR_Adjusted"] = (
    check_impressions["Clicks"]
    / check_impressions["Impressions_Adjusted"])

check_impressions["ER_Adjusted"] = (
    check_impressions["Total_Interactions"]
    / check_impressions["Impressions_Adjusted"])

check_impressions[[
        "Content_ID",
        "Impressions",
        "Impressions_Adjusted",
        "CTR",
        "CTR_Adjusted",
        "Engagement_Rate_Impressions",
        "ER_Adjusted" ]]
```

```python
df_sosmed.loc[check_impressions.index,
    "Impressions"] = (
    check_impressions["Impressions_Adjusted"]
    .round()
    .astype(int))
```

### Check dan Handling Kolom Likes

```python
check_likes = df_sosmed[
    df_sosmed["Likes"] > df_sosmed["Reach"]].copy()

check_likes[[
        "Content_ID",
        "Reach",
        "Impressions",
        "Likes",
        "Comments",
        "Shares",
        "Saves",
        "Engagement_Rate_Reach",
        "Engagement_Rate_Impressions" ]]
```

```python
check_likes["Total_Interaction_From_ER"] = (
    check_likes["Engagement_Rate_Reach"]
    * check_likes["Reach"]).round()

check_likes["Likes_Corrected"] = (
    check_likes["Total_Interaction_From_ER"]
    - check_likes["Comments"]
    - check_likes["Shares"]
    - check_likes["Saves"])

check_likes[[
        "Content_ID",
        "Likes",
        "Total_Interaction_From_ER",
        "Comments",
        "Shares",
        "Saves",
        "Likes_Corrected"]]
```

```python
df_sosmed["Likes_Original"] = df_sosmed["Likes"]
```

```python
likes_error = df_sosmed["Likes"] > df_sosmed["Reach"]

total_interaction_corrected = (
    df_sosmed.loc[likes_error, "Engagement_Rate_Reach"]
    * df_sosmed.loc[likes_error, "Reach"]).round()

likes_corrected = (
    total_interaction_corrected
    - df_sosmed.loc[likes_error, "Comments"]
    - df_sosmed.loc[likes_error, "Shares"]
    - df_sosmed.loc[likes_error, "Saves"])
```

```python
df_sosmed[df_sosmed["Likes_Original"] != df_sosmed["Likes"]][
    ["Content_ID", "Likes_Original", "Likes"]]
```

```python
df_sosmed.loc[check_likes.index,
    "Likes"] = (
    check_likes["Likes_Corrected"]
    .round()
    .astype(int))
```

### Check dan Handling Kolom Clicks

```python
check_clicks = df_sosmed[
    df_sosmed["Clicks"] > df_sosmed["Impressions"]].copy()

check_clicks[[
        "Content_ID",
        "Impressions",
        "Clicks",
        "CTR"]]
```

```python
check_clicks["Clicks_From_CTR"] = (
    check_clicks["CTR"]
    * check_clicks["Impressions"]).round()

check_clicks["Clicks_Minus_Impressions"] = (
    check_clicks["Clicks"]
    - check_clicks["Impressions"])

check_clicks[[
        "Content_ID",
        "Impressions",
        "Clicks",
        "CTR",
        "Clicks_From_CTR",
        "Clicks_Minus_Impressions"]]
```

```python
df_sosmed["Clicks_Original"] = df_sosmed["Clicks"]
```

```python
clicks_error = df_sosmed["Clicks"] > df_sosmed["Impressions"]

df_sosmed.loc[clicks_error, "Clicks"] = (
    df_sosmed.loc[clicks_error, "CTR"]
    * df_sosmed.loc[clicks_error, "Impressions"]).round().astype(int)
```

### Total Interactions
Total Interactions = Likes + Comments + Shares + Saves

Digunakan untuk menghitung seluruh interaksi utama yang diterima konten.

```python
df_sosmed["Total_Interactions"] = (
    df_sosmed["Likes"]
    + df_sosmed["Comments"]
    + df_sosmed["Shares"]
    + df_sosmed["Saves"]).astype(int)
```

```python
rename_columns = {}

if "CTR" in df_sosmed.columns:
    rename_columns["CTR"] = "CTR_Source"

if "Engagement_Rate_Reach" in df_sosmed.columns:
    rename_columns["Engagement_Rate_Reach"] = "IPR_Source"

if "Engagement_Rate_Impressions" in df_sosmed.columns:
    rename_columns["Engagement_Rate_Impressions"] = "IPI_Source"

df_sosmed.rename(columns=rename_columns, inplace=True)
```

```python
df_sosmed.columns
```

### Engagement Rate
ER = Total Interactions / Follower Count × 100%

Digunakan untuk mengukur interaksi konten dibandingkan jumlah followers akun.


```python
# Engagement Rate berdasarkan followers
df_sosmed["ER"] = (
    df_sosmed["Total_Interactions"]
    / df_sosmed["Follower_Count"])
```

### Interaction per Reach
IPR = Total Interactions / Reach × 100%

Digunakan untuk mengukur persentase interaksi dari pengguna unik yang melihat konten.

```python
# Interaction per Reach
df_sosmed["IPR"] = (
    df_sosmed["Total_Interactions"]
    / df_sosmed["Reach"])
```

### Interaction per Impression
IPI = Total Interactions / Impressions × 100%

Digunakan untuk mengukur interaksi dibandingkan seluruh frekuensi penayangan konten.

```python
# Interaction per Impression
df_sosmed["IPI"] = (
    df_sosmed["Total_Interactions"]
    / df_sosmed["Impressions"])
```

### Click-Through Rate
CTR = Clicks / Impressions × 100%

Digunakan untuk mengukur persentase impressions yang menghasilkan klik.

```python
# Click-Through Rate
df_sosmed["CTR"] = (
    df_sosmed["Clicks"]
    / df_sosmed["Impressions"])
```
